```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine
import pymssql


def get_connection():
    return pymssql.connect(
        server='localhost',
        port='1433',
        user='sa',
        password='Poor@b19',
        database='AQI'
    )

engine = create_engine('mssql+pymssql://', creator=get_connection)

with engine.connect() as conn:
    print(" Connected to MS SQL Server!")
```

Loading historical_clean.csv file in to the SQL Server

```python
## Load the historical_clean.csv file

df = pd.read_csv('/Users/poorabagrawal/Desktop/AQI/Raw_Data/historical_clean.csv', encoding='latin-1')
df.columns = df.columns.str.replace(' ', '_')
print(f" Dataset loaded!")
print(f"   Rows    : {len(df):,}")
print(f"   Columns : {df.columns.tolist()}")
```

```python
# Pushed the data to SQL Server
df.to_sql(
    name='air_quality_historical',
    con=engine,
    if_exists='replace',
    index=False,
    schema='dbo'
)
print("Data pushed to SQL Server!")
print("   Table : dbo.air_quality_historical")
print(f"   Rows  : {len(df):,}")
```
